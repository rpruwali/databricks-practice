```python
%sql
CREATE OR REPLACE TABLE workspace.default.patients (
  id INT,
  name STRING,
  city STRING
);

INSERT INTO workspace.default.patients VALUES
  (1, 'Asha', 'Raleigh'),
  (2, 'Ben', 'Durham'),
  (3, 'Chen', 'Cary');

SELECT * FROM workspace.default.patients;
```

```python
%sql
UPDATE workspace.default.patients SET city = 'Apex' WHERE id = 2;
DELETE FROM workspace.default.patients WHERE id = 3;

SELECT * FROM workspace.default.patients;
```

```python
%sql
DESCRIBE HISTORY workspace.default.patients;
```

```python
%sql
SELECT * FROM workspace.default.patients VERSION AS OF 1;
```

```python
%sql
MERGE INTO workspace.default.patients t
USING (SELECT 2 AS id, 'Ben' AS name, 'Morrisville' AS city
       UNION ALL SELECT 4, 'Dana', 'Wake Forest') s
ON t.id = s.id
WHEN MATCHED THEN UPDATE SET *
WHEN NOT MATCHED THEN INSERT *;

SELECT * FROM workspace.default.patients;
```
